## 2.1 The `print()` Function

The `print()` function displays whatever is inside the parentheses on the screen. It is the first tool you will use to see the result of your program.

In this section you will learn how to print **text, numbers, calculations, and multiple values together**.

### Printing Text

Text must be enclosed in **quotation marks**, either double (`" "`) or single (`' '`).

````python
print("Welcome to Python Course Workshop!")
print("Python is fun to learn!")
````

### Printing Numbers

Numbers are written **without** quotation marks. Both integers (whole numbers) and decimals are accepted.

````python
print(2024)
print(3.14)
````

### Printing Calculations

Python evaluates the expression first and then prints the **result**.

````python
print(10 + 5)    # Addition
print(20 - 8)    # Subtraction
print(4 * 3)     # Multiplication
print(10 / 2)    # Division
print(5 ** 2)    # Exponentiation (5 squared)
````

**Compare:** `print(5 + 3)` prints `8`, whereas `print("5 + 3")` (with quotation marks) prints the text exactly as written: `5 + 3`. Try both.

````python
print(5 + 3)
print("5 + 3")
````

### Printing Multiple Values

Separate values with a **comma** `,`. Python automatically inserts a **space** between them.

````python
print("Welcome to", "Python Course", "Workshop!")
print("Year:", 2025, "Session:", 1)
````

### Formatting Output

Special sequences that begin with a backslash `\` control the layout of the text:

- `\n` is a **new line**: output continues on the next line.
- `\t` is a **tab**: it inserts a wide horizontal space, which is useful for aligning columns.

````python
print("Python Course Workshop\nPython Basics")   # New line
````

````python
print("Name\tRole\tLevel")
print("Alice\tLeader\tExpert")
print("Wei\tMember\tIntermediate")
print("Carlos\tMember\tBeginner")
````

**Bonus.** You can also control the **separator** between values with `sep`, and the **ending** of the line with `end`.

````python
print("2025", "10", "22", sep="-")      # custom separator
print("Loading", end="...")             # no new line at the end
print("Done!")
````

## 2.2 Comments

**Comments** are notes that a programmer writes inside the code to explain or clarify the program. They are **meant for humans** and are **not executed**: Python ignores them completely.

**Why use comments?**

- They explain complex code and make it easier to understand.
- They describe the algorithm or logic being used.
- They improve **readability**, for you and for others, even months later.

### Types of Comments

```
Comments
 ├── Single-line  →  starts with  #
 └── Multi-line   →  between  ''' ... '''
```

### Single-Line Comments: `#`

Everything after a `#` on a line is treated as a comment.

**Tip:** In Colab, select a line and press **Ctrl + /** to comment or uncomment it quickly.

````python
# Printing workshop details
print("Workshop Name:\t Python Basics")
print("Course Name:\t Python Course")   # a comment can also go at the end of a line
````

### Multi-Line Comments: `''' text '''`

Longer comments can be placed between **three quotation marks** `'''` ... `'''`.

*Technical note:* this is actually a string that is not assigned to any variable, so Python simply ignores it. It is a common way to write a multi-line comment. You can also place a `#` at the start of each line.

````python
'''
This part prints
the number of participants and duration
'''
print("Participants:\t", 46)
print("Duration:\t", 7, "hours")
````

### Combined Example

````python
# Printing workshop details
print("Workshop Name:\t Python Basics")
print("Course Name:\t Python Course")

'''
This part prints
the number of participants and duration
'''
print("Participants:\t", 46)
print("Duration:\t", 7, "hours")
````

## 2.3 Variables and Data Types

### What Is a Variable?

A **variable** is a **container** for storing data. Think of it as a **labeled box** that holds a piece of information such as a number or a piece of text. Variables let us store values and then use or modify them later.

In the example `age = 60`, the variable name is `age` and its value is `60`. The value is stored in the computer's memory (**RAM**).

<p align="center"></p>

<p align="center"><sub>A variable stored in memory (RAM)</sub></p>

### How Variables Work in Python

- In Python you **do not declare the data type** explicitly. Python determines it automatically from the value.
- You create a variable by writing a **name**, then the `=` sign, then the **value**.

````python
workshop_name = "Python Basics"   # Storing a string
participants = 25                 # Storing an integer
duration = 3.5                    # Storing a float
is_online = True                  # Storing a boolean

print(workshop_name)
print(participants)
print(duration)
print(is_online)
````

### Data Types

Every value has a **type**. The table below summarizes the basic types. You can find the type of any value using the `type()` function.

| Type | Name | Example | Usage |
|:---|:---:|:---|:---|
| Integer | `int` | `x = 124` | Numbers without a decimal point |
| Float | `float` | `x = 124.56` | Numbers with a decimal point |
| String | `str` | `x = "Hello world"` | Text |
| Boolean | `bool` | `x = True` or `x = False` | Conditions and logic |
| NoneType | `None` | `x = None` | Representing "no value" or an empty variable |

````python
x = 124
print(x, "->", type(x))

x = 124.56
print(x, "->", type(x))

x = "Hello world"
print(x, "->", type(x))

x = True
print(x, "->", type(x))

x = None
print(x, "->", type(x))
````

### Rules for Naming Variables

1. A name must start with a **letter** (`a-z`, `A-Z`) or an **underscore** `_`.
2. A name **cannot** start with a **digit**.
3. A name can contain **letters, digits, and underscores**.
4. Names are **case-sensitive**: `course` and `Course` are two different variables.
5. A name cannot be a reserved keyword of the language (for example `if`, `for`, `class`).

| Name | Valid? | Reason |
|:---|:---:|:---|
| `course_name = "Python Course"` | Valid | starts with a letter |
| `_members = 30` | Valid | starts with an underscore |
| `1workshop = "Python"` | Invalid | cannot start with a digit |
| `workshop_2025 = "AI, Robotics, PCB Design Workshops Soon"` | Valid | letters, digits, and `_` are allowed |
| `course` and `Course` | Valid | different variables (case-sensitive) |

````python
course_name = "Python Course"     # Valid
_members = 30                     # Valid
workshop_2025 = "AI, Robotics, PCB Design Workshops Soon"   # Valid

print(course_name, _members, workshop_2025)
````

### Operators on Strings and Numbers

The same operator can behave differently depending on the type of its operands:

- With **strings**, `+` **joins** the text together (*concatenation*).
- With **numbers**, `+` **adds** and `*` **multiplies**.

````python
# 1) Strings
text_1 = "python"
text_2 = "course"
print(text_1 + text_2)          # Output: pythoncourse
print(text_1 + " " + text_2)    # Output: python course
````

````python
# 2) Numbers
number_1 = 5
number_2 = 5
print(number_1 + number_2)      # Output: 10
print(number_1 * number_2)      # Output: 25
````

### Common Error: Mixing Text and Numbers

You **cannot** add a string and a number directly. Python cannot tell whether you intend addition or concatenation, so it raises a `TypeError`.

*Note:* the next cell uses `try / except` **only** so that the notebook does not stop at the error. You will study it later. What matters here is that you read the error message.

````python
age = 22
try:
    print("My age is " + age)          # This line causes an error
except TypeError as error:
    print("Error:", error)
````

#### Solution: convert the number to a string with `str()` (*casting*)

**Casting** means changing the type of a value. Here we convert the number to a string with `str()`, which makes concatenation possible.

````python
age = 22
print("My age is " + str(age))         # Correct
````

## 2.5 Sequence and Mapping Data Types

So far, each variable has held a single value. Often we want to store a **collection** of values in one variable. Python provides three fundamental types for this: **lists**, **tuples**, and **dictionaries**. We begin with the most important one.

### 1. Lists

- **Ordered:** every element has a fixed position.
- **Mutable:** elements can be added, removed, and modified.
- Written between **square brackets** `[]`.

````python
fruits = ["apple", "orange", "tomato", "banana"]
print(fruits)
````

### Lists: Indexing

List numbering may seem unintuitive at first: the **first element has index 0**, not 1.

| Index | 0 | 1 | 2 | 3 |
|:---|:---:|:---:|:---:|:---:|
| **Element** | `"apple"` | `"orange"` | `"tomato"` | `"banana"` |
| **Negative index** | -4 | -3 | -2 | -1 |

````python
fruits = ["apple", "orange", "tomato", "banana"]
print(fruits[0])     # first element
print(fruits[1])     # second element
````

**Tip.** A negative index counts from the end: `fruits[-1]` is the **last** element.

````python
print(fruits[-1])    # last element
print(fruits[-2])    # second to last
````

> **Warning.** What if we request an index that does not exist? The list has only 4 elements (indices 0 to 3), so requesting `fruits[4]` raises `IndexError: list index out of range`.

````python
fruits = ["apple", "orange", "tomato", "banana"]
try:
    print(fruits[4])
except IndexError as error:
    print("IndexError:", error)
````

### Lists: Size

Always be aware of the **size** of your data. The `len()` function returns the **length** of any sequence.

````python
fruits = ["apple", "orange", "tomato", "banana"]
print(len(fruits))     # Output: 4
````

### Lists: Adding, Removing, and Modifying Elements

#### 1. Adding elements

- `append()` adds an element **at the end** of the list.
- `insert()` adds an element **at a specific position** (index first, then value).

````python
fruits = ["apple", "orange", "tomato", "banana"]

# Adding a new fruit at the end
fruits.append("mango")
print(fruits)      # ['apple', 'orange', 'tomato', 'banana', 'mango']

# Adding a fruit at index 2
fruits.insert(2, "grape")
print(fruits)      # ['apple', 'orange', 'grape', 'tomato', 'banana', 'mango']
````

#### 2. Removing elements

- `remove()` deletes a specific **value**.
- `pop()` deletes an element **by index**. If no index is given, it removes the **last** element.

````python
# By value
fruits = ["apple", "orange", "tomato", "banana"]

fruits.remove("orange")
print(fruits)      # ['apple', 'tomato', 'banana']
````

````python
# By index
fruits = ["apple", "orange", "tomato", "banana"]

fruits.pop(1)
print(fruits)      # ['apple', 'tomato', 'banana']
````

#### 3. Modifying elements

Change an element by assigning a new value to it **using its index**.

````python
fruits = ["apple", "orange", "tomato", "banana"]

# Changing "tomato" to "grape"
fruits[2] = "grape"
print(fruits)      # ['apple', 'orange', 'grape', 'banana']
````

### 2. Tuples

- **Ordered**
- **Immutable:** they **cannot** be changed after creation.
- Written between **parentheses** `()`.

````python
colors = ("red", "green", "blue")
print(colors)
print(colors[0])

# Tuples are immutable: changing an element is not allowed
try:
    colors[0] = "yellow"
except TypeError as error:
    print("Error:", error)
````

### 3. Dictionaries

- Store **key-value pairs**.
- Written between **curly braces** `{}`.
- A value is read through its **key**, not through a numeric position.

````python
student = {"name": "Maria", "age": 20, "city": "Madrid"}

print(student)
print(student["name"])       # read a value by its key

student["age"] = 21          # update a value
student["level"] = "Beginner"  # add a new key-value pair
print(student)
````

## 3.1 Operators

**Operators** are symbols that perform operations on variables and values. For example, the `+` operator adds two values.

In this course we cover four groups of operators:

1. **Arithmetic operators**
2. **Assignment operators**
3. **Comparison operators**
4. **Logical operators**

### Arithmetic Operators

| Operator | Name | Example |
|:---:|:---|:---|
| `+` | Addition | `x + y` |
| `-` | Subtraction | `x - y` |
| `*` | Multiplication | `x * y` |
| `/` | Division | `x / y` |
| `%` | Modulus (remainder) | `x % y` |
| `**` | Exponentiation | `x ** y` |
| `//` | Floor division | `x // y` |

````python
x = 10
y = 3

print("x + y  =", x + y)     # 13
print("x - y  =", x - y)     # 7
print("x * y  =", x * y)     # 30
print("x / y  =", x / y)     # 3.3333333333333335
print("x % y  =", x % y)     # 1   (remainder of 10 / 3)
print("x ** y =", x ** y)    # 1000 (10 to the power 3)
print("x // y =", x // y)    # 3   (division rounded down to a whole number)
````

**Tip.** The `%` operator is very useful. For example, `n % 2 == 0` means that `n` is **even**. We will use it with conditions shortly.

### Assignment Operators

Assignment operators store a value in a variable. A compound operator such as `+=` is a **shorthand**: `x += 3` means `x = x + 3`.

| Operator | Example | Same as |
|:---:|:---:|:---:|
| `=` | `x = 5` | `x = 5` |
| `+=` | `x += 3` | `x = x + 3` |
| `-=` | `x -= 3` | `x = x - 3` |
| `*=` | `x *= 3` | `x = x * 3` |
| `/=` | `x /= 3` | `x = x / 3` |
| `%=` | `x %= 3` | `x = x % 3` |
| `//=` | `x //= 3` | `x = x // 3` |
| `**=` | `x **= 3` | `x = x ** 3` |
| `&=` `\|=` `^=` `>>=` `<<=` | `x &= 3` ... | bitwise operators (advanced) |
| `:=` | `print(x := 3)` | `x = 3`, then `print(x)` |

````python
x = 10
x += 3
print("x += 3  ->", x)     # 13

x = 10
x -= 3
print("x -= 3  ->", x)     # 7

x = 10
x *= 3
print("x *= 3  ->", x)     # 30

x = 10
x /= 4
print("x /= 4  ->", x)     # 2.5

x = 10
x %= 3
print("x %= 3  ->", x)     # 1

x = 10
x //= 3
print("x //= 3 ->", x)     # 3

x = 10
x **= 2
print("x **= 2 ->", x)     # 100

print(y := 3)              # assigns 3 to y AND prints it (the "walrus" operator)
````

### Comparison Operators

Comparison operators compare two values and **always** return `True` or `False`.

> **Warning.** A common mistake: `=` is used for **assignment** (storing a value), whereas `==` is used for **comparison** (testing whether two values are equal).

| Operator | Name | Example |
|:---:|:---|:---:|
| `==` | Equal | `x == y` |
| `!=` | Not equal | `x != y` |
| `>` | Greater than | `x > y` |
| `<` | Less than | `x < y` |
| `>=` | Greater than or equal to | `x >= y` |
| `<=` | Less than or equal to | `x <= y` |

````python
x = 5
y = 10

print("x == y :", x == y)    # False
print("x != y :", x != y)    # True
print("x >  y :", x > y)     # False
print("x <  y :", x < y)     # True
print("x >= y :", x >= y)    # False
print("x <= y :", x <= y)    # True
````

### Logical Operators

Logical operators combine several conditions into a single one:

- `and` returns `True` if **both** conditions are true.
- `or` returns `True` if **at least one** condition is true.
- `not` **reverses** the result (`True` becomes `False` and vice versa).

| Operator | Description | Example |
|:---:|:---|:---|
| `and` | Returns True if both statements are true | `x < 5 and x < 10` |
| `or` | Returns True if one of the statements is true | `x < 5 or x < 4` |
| `not` | Reverses the result: returns False if the result is true | `not(x < 5 and x < 10)` |

````python
x = 3

print(x < 5 and x < 10)        # True  (both are true)
print(x < 5 or x < 4)          # True  (at least one is true)
print(not(x < 5 and x < 10))   # False (reverses True)
````

### Example 1: A Simple `while` Loop

We start with `count = 1` and increase it on each iteration. When it becomes greater than 5, the condition becomes false and the loop stops.

````python
count = 1                                    # Starting value

while count <= 5:                            # Condition
    print("Python Course Workshop", count)
    count += 1                               # Increment to avoid an infinite loop
````

### Example 2: Avoiding an Infinite Loop

> **Warning.** If the condition **never changes**, the loop becomes **infinite** and runs forever.

The following example must **not** be run. The condition `num > 0` is always true because `num` never changes:

```python
num = 1
while num > 0:                       # Always True: infinite loop
    print("This will run forever!")
```

**If you get stuck in an infinite loop in Colab:** click the Stop button next to the cell, or use **Ctrl + M, then I** (Interrupt execution).

**The fix:** always make sure that something inside the loop **changes the condition**, so that it eventually becomes false.

````python
num = 3
while num > 0:
    print("This will run", num, "more time(s)")
    num -= 1                         # num changes, so the condition becomes False at 0
print("Loop finished safely!")
````

### Stopping a `while` Loop with `break`

````python
num = 1
while num <= 10:
    if num == 5:
        break                # Stops the loop when num is 5
    print("Number:", num)
    num += 1
````

# Part 05: Functions

## 5.1 What Is a Function in Python?

> A **function** is a set of statements that together perform a specific task.

A function groups instructions that carry out a **specific task** and gives them a **name**, so that we can call them whenever we need.

Think of a coffee machine: you give it **inputs** (beans and water), it performs an **operation**, and it gives you a **result** (coffee). In fact, we have been using built-in functions from the very beginning: `print()`, `input()`, `len()`, `range()`, `int()`, and others.

<p align="center"></p>

### Advantages of Functions

1. **Modularity.** Functions divide a program into parts. A large problem can be broken into smaller problems, each solved by its own function.
2. **Reusability.** Once a function is defined, you can **call it anywhere** without copying and pasting its logic. You never need to write the same code repeatedly.
3. **Maintainability.** Because functions divide the program, it is easier to maintain. To change the behavior later, you only need to **update the function**, without touching the rest of the code.

### Anatomy of a Function

- `def` is the keyword that means "define a function".
- The **function name** is chosen by you (following the same rules as variable names).
- The **parentheses** `()` contain the **parameters**, if any.
- A **colon** `:` is followed by an **indented block of code**.
- `return` (optional) sends a **result** back from the function.

**Tip.** A *parameter* is the name used inside the function definition, whereas an *argument* is the actual value passed when the function is called.

```python
def function_name(parameter1, parameter2):   # 1) definition
    # code of the function
    return result                            # 2) (optional) give a result back

function_name(argument1, argument2)          # 3) calling the function
```

## 5.2 Syntax of Functions in Python

We will look at five kinds of functions, from the simplest to the most commonly used.

### 1. Defining and Calling a Function

First we **define** the function with `def` (this prints nothing by itself). Then we **call** it by writing its name followed by parentheses, which executes the code inside. A function can be called as many times as needed.

````python
def greet():
    print("Hello, welcome to Python Course!")

# Calling the function
greet()
greet()      # we can call it as many times as we want
````

### 2. A Function with Parameters

**Parameters** make a function flexible: we pass different values on each call and obtain different results.

````python
def greet_student(name):
    print("Hello, " + name + "! Welcome to Python Course.")

# Calling with different arguments
greet_student("Priya")
greet_student("Lucas")
````

### 3. A Function with a Return Value

`return` **sends a result back** to the place from which the function was called. We can then **store it in a variable** and use it later, instead of merely printing it.

````python
def add(a, b):
    return a + b             # Returns the sum

# Storing and using the returned value
result = add(10, 5)
print("The sum is:", result)
````

**Tip: `print` versus `return`.** `print` only **displays** a value on the screen. `return` **hands** the value to the program so that it can be used further (in a calculation, a variable, a condition, and so on).

### 4. A Function with a Default Parameter Value

A parameter can be given a **default value**. It is used when **no value is passed** at the time of the call.

````python
def workshop(topic="AI"):
    print("Today's workshop is about", topic)

# Calling without an argument (uses the default)
workshop()

# Calling with an argument
workshop("Robotics")
````

### 5. A Function with Multiple Parameters

A function can have **more than one parameter**, separated by commas. Values are passed **in the same order**.

````python
def multiply(x, y):
    return x * y

print(multiply(3, 4))     # Output: 12
print(multiply(5, 2))     # Output: 10
````

````python
#@title Solution (attempt the exercise first) {display-mode: "form"}
readings = [21.5, 23.0, 22.4, 26.8, 31.2, 28.9, 24.1]
threshold = 25


def average(values):
    return sum(values) / len(values)


highest = readings[0]
lowest = readings[0]
count_above = 0

for value in readings:
    if value > highest:
        highest = value
    if value < lowest:
        lowest = value
    if value > threshold:
        count_above += 1

print("Highest:", highest)                                 # 31.2
print("Lowest:", lowest)                                   # 21.5
print("Readings above", threshold, ":", count_above)       # 3
print("Average:", round(average(readings), 2))             # 25.41
````
